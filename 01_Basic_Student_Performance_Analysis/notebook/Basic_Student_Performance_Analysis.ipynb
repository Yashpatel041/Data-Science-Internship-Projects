# Basic Student Performance Analysis

## Mini Project 1

This project analyzes the performance of students using basic Python programming concepts.

### Objectives

The main objectives of this project are:

- Store student names and their grades using Python data structures.
- Calculate the average grade of each student.
- Determine whether a student has passed or failed.
- Handle students with missing or invalid grade information.
- Find the highest and lowest performing students.
- Use Python functions to perform different calculations.
- Demonstrate conditional statements and loops.
- Demonstrate string operations.
- Use `continue` to skip invalid student entries.
- Use `break` to stop the analysis when a specified condition is reached.

### Python Concepts Used

- Lists
- Dictionaries
- Strings
- Arithmetic operators
- Conditional statements
- `for` loops
- User-defined functions
- `continue`
- `break`

```python
students = [
    {"name": "Alice", "grades": [78, 85, 90]},
    {"name": "Bob", "grades": [55, 60, 58]},
    {"name": "Charlie", "grades": [92, 88, 95]},
    {"name": "David", "grades": [30, 40, 25]},
    {"name": "Eva", "grades": [100, 100, 98]},
    {"name": "Invalid", "grades": []}
]
```

```python
# Function to calculate average
def calculate_average(grades):
    if len(grades) == 0:
        return 0
    return sum(grades) / len(grades)


# Function to determine pass/fail
def is_pass(average, threshold=50):
    return average >= threshold


# Function to get highest and lowest average student
def get_extreme_students(student_list):
    highest = None
    lowest = None
    highest_avg = float('-inf')
    lowest_avg = float('inf')

    for student in student_list:
        avg = calculate_average(student['grades'])

        if avg > highest_avg:
            highest = student['name']
            highest_avg = avg

        if avg < lowest_avg:
            lowest = student['name']
            lowest_avg = avg

    return highest, highest_avg, lowest, lowest_avg
```

```python
print("===== Student Performance Analysis =====\n")

for student in students:
    name = student['name'].title()
    grades = student['grades']

    # Skip invalid entries using continue
    if not grades:
        print(f"Skipping {name} due to missing grades.\n")
        continue

    average = calculate_average(grades)
    status = "Pass" if is_pass(average) else "Fail"

    print(f"Name: {name}")
    print(f"Grades: {grades}")
    print(f"Average: {average:.2f}")
    print(f"Status: {status}\n")

    # Example break condition: stop if someone scores a perfect average
    if average == 100:
        print("Perfect scorer found! Stopping analysis.\n")
        break
```

```python
# Filter out invalid students before finding highest and lowest
valid_students = [s for s in students if s['grades']]

# Get highest and lowest performers
high_name, high_avg, low_name, low_avg = get_extreme_students(valid_students)

# Display the summary
print("===== Summary =====")
print(f"Highest Performer: {high_name} with {high_avg:.2f}")
print(f"Lowest Performer: {low_name} with {low_avg:.2f}")
```
